```python
import time

async def plus(x,y, sleep=1):
    time.sleep(sleep)
    return x+y
```

```python

one = plus(3,2, 10)
print(f"One: {one}")
two = plus(5,10, 5)
print(f"Two: {two}")
three = plus(5,3, 20)
print(f"Three: {three}")
four = plus(5,1,1)
print(f"Four: {four}")
```

```python
import asyncio
results = asyncio.gather(plus(3,2, 10), plus(5,10, 5), plus(5,3, 20), plus(5,1,1))
```
